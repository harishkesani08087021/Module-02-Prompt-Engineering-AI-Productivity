## Day 6 – Prompt Engineering Fundamentals

### Key concepts
- **Zero-shot prompting:** Ask a model to perform a task without giving examples.
- **Few-shot prompting:** Give examples that show the desired pattern before the actual task.
- **Role:** The perspective or expertise the model should adopt.
- **Context:** Background information needed for the task.
- **Instruction:** The action the model should perform.
- **Constraints:** Rules such as length, tone, scope, or content to avoid.

A useful prompt structure is: **Role + Context + Instruction + Constraints + Output format**.

```python
# Zero-shot prompt
zero_shot_prompt = "Explain prompt engineering to a beginner in three bullet points."
print(zero_shot_prompt)

# Few-shot prompt: examples demonstrate the expected transformation
few_shot_prompt = '''
Convert each message into a sentiment label.

Example 1:
Message: "The support team solved my issue quickly."
Label: Positive

Example 2:
Message: "The product arrived broken."
Label: Negative

Now classify:
Message: "The delivery was on time and the packaging was good."
Label:
'''
print(few_shot_prompt)
```

## Day 7 – Structured Outputs: JSON, Summarization, Classification, Generation

Structured outputs make model responses easier to read and process. JSON uses key-value pairs and must follow valid syntax.

```python
import json

# Example of a valid structured JSON response
sample_response = {
    "topic": "Prompt Engineering",
    "summary": "Writing clear instructions to guide an AI model.",
    "difficulty": "Beginner",
    "key_points": [
        "Give clear instructions",
        "Provide relevant context",
        "Specify the output format"
    ]
}

json_text = json.dumps(sample_response, indent=2)
print(json_text)

# Validate that the JSON can be parsed
parsed = json.loads(json_text)
print("\\nJSON is valid:", isinstance(parsed, dict))
```

### Reusable prompts for common tasks

**Summarization prompt**
> Summarize the text below in 5 bullet points. Keep the important facts, use simple English, and do not add information that is not in the text. Text: [paste text]

**Classification prompt**
> Classify each item into one of these categories: [categories]. Return only valid JSON with keys `item` and `category`. If uncertain, use `Needs review`. Items: [paste items]

**Generation prompt**
> Create [content type] for [audience] about [topic]. Use a [tone] tone, keep it within [length], and return it in [format].

```python
# Python helper: create a prompt from reusable fields
def build_prompt(role, context, task, constraints, output_format):
    return f"""Role: {role}
Context: {context}
Task: {task}
Constraints: {constraints}
Output format: {output_format}"""

example_prompt = build_prompt(
    role="Study assistant",
    context="A beginner learning Python",
    task="Explain a for loop",
    constraints="Use simple English and one short example",
    output_format="Heading, explanation, code"
)
print(example_prompt)
```

## Day 8 – AI-assisted Coding and Debugging

AI can help explain code, suggest fixes, generate test cases, and improve readability. Always run and review suggested code; AI-generated code may contain errors or security issues.

### Debugging prompt template
> Act as a Python debugging assistant. Explain the error in simple English. Identify the likely cause, suggest the smallest safe correction, and provide a corrected code snippet. Do not change unrelated parts. Code: [paste code] Error: [paste full error]

```python
# Deliberately incorrect example
def calculate_average(numbers):
    return sum(numbers) / len(number)

# Try to identify the bug before reading the correction.
print("Bug: the function uses 'number' instead of 'numbers' in len().")

# Corrected version with an empty-list check
def calculate_average_fixed(numbers):
    if not numbers:
        return None
    return sum(numbers) / len(numbers)

print("Average:", calculate_average_fixed([10, 20, 30]))
print("Empty list:", calculate_average_fixed([]))
```

### Coding assistance prompt templates

## Day 9 – Prompt Evaluation and Reusable Templates

A good prompt should be evaluated consistently rather than judged only by whether the answer sounds fluent.

### Prompt evaluation criteria
1. **Relevance:** Does the response answer the actual request?
2. **Accuracy:** Are factual claims correct and verifiable?
3. **Completeness:** Are all requested parts included?
4. **Clarity:** Is the response easy to understand?
5. **Format compliance:** Does it follow the requested structure?
6. **Safety:** Does it avoid harmful, biased, or private information?

```python
# Simple prompt testing report template
prompt_test_report = {
    "test_name": "Beginner explanation test",
    "prompt_version": "v1",
    "task": "Explain context windows",
    "model_used": "Record the model you actually tested",
    "criteria_scores_out_of_5": {
        "relevance": None,
        "accuracy": None,
        "completeness": None,
        "clarity": None,
        "format_compliance": None,
        "safety": None
    },
    "observations": "Write what worked and what should be improved.",
    "next_change": "Make one specific improvement to the prompt."
}
print(json.dumps(prompt_test_report, indent=2))
```

## Day 10 – AI Productivity Toolkit

The toolkit below contains reusable workflows for common study and workplace activities. Replace bracketed text with your own content. Review all AI outputs before sending or submitting them.

### 1. Summarization workflow
> Summarize the following material for [audience]. Provide: (1) a 2-sentence overview, (2) 5 key points, (3) important terms with definitions, and (4) 3 revision questions. Use only the supplied material and mention if information is missing. Material: [paste text]

### 2. Email workflow
> Draft a polite, professional email to [recipient/role] about [purpose]. Include a clear subject, brief context, the specific request, and a courteous closing. Keep it under [word limit]. Do not invent dates, names, or facts.

### 3. Resume workflow
> Review my resume for a [target role]. Suggest improvements to clarity, action verbs, and relevance. Do not invent qualifications, projects, skills, or achievements. Separate suggested edits from facts that need my confirmation. Resume text: [paste resume text]

### 4. Coding workflow
> Help me implement [task] in [programming language]. First outline the approach, then provide readable code, explain important lines, and include test cases for normal, boundary, and invalid inputs. State assumptions and possible limitations.

### 5. Research assistance workflow
> Help me research [topic]. Provide a beginner-friendly overview, key concepts, and questions to investigate further. Distinguish established facts from uncertain claims. Provide verifiable source names or links when available; never fabricate citations. I will independently verify sources.

```python
# Store reusable productivity prompts in a Python dictionary
productivity_toolkit = {
    "summarization": "Summarize [material] for [audience] in 5 key points. Do not add unsupported facts.",
    "email": "Draft a polite email to [recipient] about [purpose], under [word limit]. Do not invent details.",
    "resume": "Review my resume for [role]. Suggest improvements without inventing experience.",
    "coding": "Help implement [task] in [language]. Include explanation and test cases.",
    "research": "Explain [topic], distinguish facts from uncertainty, and suggest verifiable sources."
}

for workflow, template in productivity_toolkit.items():
    print(f"\\n{workflow.upper()} PROMPT:")
    print(template)
```
